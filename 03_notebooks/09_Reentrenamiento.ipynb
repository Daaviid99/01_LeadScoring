```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
from pathlib import Path
import cloudpickle

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer
from sklearn.compose import make_column_transformer
from sklearn.pipeline import make_pipeline
```

# 01. Importación
---

## 1.1. Importación de datos

```python
# Definir rutas
data_path = Path('../02_datos/01_Originales')
nombre_fichero = 'Leads.csv'

# Importar Leads.csv
leads_file = data_path / nombre_fichero

df = pd.read_csv(leads_file, sep=';', encoding='utf-8', index_col='id')

print(f'✅ Importación completada')
print(f'Shape: {df.shape}')
```

# 2. Transformaciones y modificaciones

## 2.1. Duplicados y EDA

```python
# Duplicados y Filtrado por EDA
df.drop_duplicates(inplace=True)

# Filtrado por EDA
df = df.loc[
    (df.no_llamar != 'OTROS') & 
    (df.no_enviar_email != 'Yes') & 
    (df.ult_actividad != 'Email Bounced')]
```

## 2.2. Dividir variables x e y 

```python
variables_finales = ['ambito',
                     'descarga_lm',
                     'ocupacion',
                     'paginas_vistas_visita',
                     'score_actividad',
                     'score_perfil', 
                     'tiempo_en_site_total', 
                     'ult_actividad',
                     'visitas_total']

target = 'compra'
```

```python
x = df[variables_finales].copy()

y = df[target].copy()
```

# 3. Pipeline de ejecución y entrenamiento

## 3.1. Pipeline de entrenamiento

```python
# Ruta pipeline final de entrenamiento

ruta_entrenamiento = '../05_modelos/'
nombre_pipe_entrenamiento = 'pipe_entrenamiento.pkl'

ruta_pipe_entrenmiento = ruta_entrenamiento + nombre_pipe_entrenamiento
```

```python
with open(ruta_pipe_entrenmiento, mode='rb') as file:
    pipe_entrenamiento = cloudpickle.load(file)

pipe_ejecucion = pipe_entrenamiento.fit(x, y)
```

## 3.2. Pipeline de ejecución

```python
# Ruta del pipeline final de ejecución
ruta_ejecucion = '../05_modelos/'
nombre_pipe_ejecucion = 'pipe_ejecucion.pkl'

ruta_pipe_ejecucion = ruta_ejecucion + nombre_pipe_ejecucion
```

```python
# Guardamos el pipeline final de ejecución

with open(ruta_pipe_ejecucion, mode='wb') as file:
    cloudpickle.dump(pipe_ejecucion, file)
```
